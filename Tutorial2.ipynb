5. Create two random numpy arrays of size n x p and p x m and then compute the dot product

```python
import numpy as np

n = 3
p = 4
m = 2

arr1 = np.random.rand(n, p)
arr2 = np.random.rand(p, m)

product = np.dot(arr1, arr2)

print("arr1: ")
print(arr1)

print("arr2: ")
print(arr2)

print("product between arr1 and arr2: ")
print(product)
```
