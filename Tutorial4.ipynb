# **Question 1: Identify the Problem**

Consider this dataset:

Student	Math	Science	English

Ali	75	80	72

Sara	88	91	85

John	65	70	68




1. Is this data in tidy format?

No

2. If not, explain which tidy-data rule is violated.

In tidy data, each column represents a single variable, and each row represents a single observation. In this dataset, the column headers (Math, Science, English) are actually values of a variable (Subject), and the scores are spread across multiple columns instead of being stored under a single Score variable.

3. Using Python/pandas, reshape the dataset so that it has the columns:
   - Student
   - Subject
   - Score

```python
import pandas as pd

data = {
    "Student": ["Ali", "Sara", "John"],
    "Math": [75, 88, 65],
    "Science": [80, 91, 70],
    "English": [72, 85, 68]
}

df = pd.DataFrame(data)

tidied_df = pd.melt(
    df,
    id_vars=["Student"],
    value_vars=["Math", "Science", "English"],
    var_name="Subject",
    value_name="Score"
)

print(tidied_df)
```

# **Question 2: Sales Data**

You are given:

Country   Q1_Sales   Q2_Sales   Q3_Sales   Q4_Sales

Malaysia    12000      15000      14000      18000

Thailand    10000      11000      13000      16000

Indonesia   14000      12500      15000      17000

1. What is the tidy-data problem with this dataset?

2. Use pandas to convert it into:

Country    Quarter    Sales

Malaysia   Q1         12000

Malaysia   Q2         15000

```python
data = {
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000]
}

df = pd.DataFrame(data)

tidied_df = pd.melt(
    df,
    id_vars=["Country"],
    value_vars=["Q1_Sales", "Q2_Sales", "Q3_Sales", "Q4_Sales"],
    var_name="Quarter",
    value_name="Sales"
)

print(tidied_df)
```


# **Question 3: Asking Questions from Data**

Consider this dataset:
data = {    "Student": ["A", "B", "C", "D", "E", "F"],    "Study_Hours": [2, 5, 3, 7, 4, 8],    "Score": [55, 72, 60, 85, 68, 90]}




Using Python:

1. What is the average study time?

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
}

df =  pd.DataFrame(data)

average_study_time = df["Study_Hours"].mean()
print(f"Average study time: {average_study_time:.2f} hours")
```

2. What is the average score?

```python
average_score = df["Score"].mean()
print(f"Average score: {average_score:.2f}")
```

3. Which student obtained the highest score?

```python
highest_score_student = df.loc[df["Score"].idxmax(), "Student"]

print(f"Student that obtained the highest score: {highest_score_student}")
```

4. Which student studied the most?

```python
highest_study_time_student = df.loc[df["Study_Hours"].idxmax(), "Student"]

print(f"Student that studied the most: {highest_study_time_student}")
```
