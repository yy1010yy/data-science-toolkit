# 6 & 8. How many columns have null values? If yes, which column, which rows

```python
missing = dataframe.isnull().sum()

print("Missing values in each column (null): ")
print(missing)

import pandas as pd

# Define columns where zero means "missing" or "invalid"
cols_to_check = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Count how many zeros exist in each column
missing_as_zeros = (dataframe[cols_to_check] == 0).sum()
print("\n\nColumns with missing values (disguised as zeros): ")
print(missing_as_zeros)

print("\n\nColumns with missing values (null): ")
print(missing[missing > 0])

print("\n\nNumber of columns with missing values: ", (missing > 0).sum())
print("\nNumber of columns with missing values (disguised as zeros): ", (missing_as_zeros > 0).sum())
```

# 7. Which column is the feature, which is the target output?
